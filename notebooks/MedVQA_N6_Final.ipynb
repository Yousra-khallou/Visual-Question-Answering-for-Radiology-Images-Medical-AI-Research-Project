# 🔬 MedVQA N6 — Visual Question Answering for Radiology Images
### Multimodal Architecture: ViT-Base/16 + BiomedBERT + Dual Prediction Heads + Grad-CAM
**Trained and Evaluated on VQA-RAD + SLAKE (~7,000 paired samples)**

---
### 📌 Research Journey & Ablation Summary
- **Benchmarked Baselines:**
  - **N1**: Majority Class, Random Guess & Question-Only (BERT-base)
  - **N2**: CNN (ResNet) + LSTM
  - **N3**: Bilinear Attention Network (BAN-8)
  - **N4**: ViLBERT / Co-attention
  - **N5/N6 (Final Model)**: Vision Transformer (ViT-Base/16) + BiomedBERT + Linear Fusion/BatchNorm with Dual Output Heads (Closed & Open) + Grad-CAM.
- **N6 Design Rationale**: Linear projection & concatenation fusion achieved superior stability and generalization compared to heavy BAN/Cross-Attention mechanisms on ~7k training pairs.

## 3. 🛠️ Utility Functions, Closed/Open Heuristics & Normalizers


```python
# =============================================================================
# UTILITAIRES
# =============================================================================
def pil_to_bytes(pil_img, quality=85):
    buf = io.BytesIO()
    pil_img.convert('RGB').save(buf, format='JPEG', quality=quality)
    return buf.getvalue()

# =============================================================================
# HEURISTIQUE CLOSED RESTREINTE
# Restreinte exprès : "chest", "lung" etc. → OPEN (pas closed)
# =============================================================================
CLOSED_KEYWORDS = {
    'yes', 'no', 'normal', 'abnormal', 'present', 'absent',
    'left', 'right', 'bilateral', 'central', 'peripheral',
    'true', 'false', 'positive', 'negative',
}

def is_closed_answer(answer: str) -> bool:
    a = answer.strip().lower()
    if a in CLOSED_KEYWORDS: return True
    if a.isdigit(): return True
    return False

# =============================================================================
# NORMALISEURS
# =============================================================================
def normalize_vqa_rad(sample):
    answer = str(sample['answer']).strip().lower()
    return {
        'image_bytes'  : pil_to_bytes(sample['image']),
        'question'     : str(sample['question']).strip(),
        'answer'       : answer,
        'answer_type'  : 'CLOSED' if is_closed_answer(answer) else 'OPEN',
        'question_type': 'OTHER',
    }

def normalize_slake(sample):
    answer = str(sample['answer']).strip().lower()
    raw_type = str(sample.get('answer_type', '')).strip().lower()
    return {
        'image_bytes'  : pil_to_bytes(sample['image']),
        'question'     : str(sample['question']).strip(),
        'answer'       : answer,
        'answer_type'  : 'CLOSED' if raw_type == 'closed' else 'OPEN',
        'question_type': str(sample.get('qid', 'OTHER')),
    }
```

## 4. 📥 Step 1: Sequential Dataset Loading (VQA-RAD + SLAKE)


```python
# =============================================================================
# ÉTAPE 1 : Chargement séquentiel (anti-OOM)
# =============================================================================
print("\n=== Chargement VQA-RAD ===")
_vqa = load_dataset("flaviagiammarino/vqa-rad")
vqa_rad_train = [normalize_vqa_rad(s) for s in _vqa['train']]
vqa_rad_test  = [normalize_vqa_rad(s) for s in _vqa['test']]
del _vqa; gc.collect()
print(f"VQA-RAD — Train: {len(vqa_rad_train)} | Test: {len(vqa_rad_test)}")

print("\n=== Chargement SLAKE ===")
_slake = load_dataset("mdwiratathya/SLAKE-vqa-english")
slake_train = [normalize_slake(s) for s in _slake['train']]
slake_val   = [normalize_slake(s) for s in _slake['validation']]
slake_test  = [normalize_slake(s) for s in _slake['test']]
del _slake; gc.collect()
print(f"SLAKE — Train: {len(slake_train)} | Val: {len(slake_val)} | Test: {len(slake_test)}")
```

## 5. 🔀 Step 2: Stratified Split (Train / Validation / Test)


```python
# =============================================================================
# ÉTAPE 2 : Validation STRATIFIÉE (correction root-cause N5 original)
# 225 VQA-RAD dans val → Val C toujours > 0% → early stopping correct
# =============================================================================
combined_train = vqa_rad_train + slake_train
combined_val   = vqa_rad_test[:225] + slake_val   # ← clé
combined_test  = vqa_rad_test[225:] + slake_test

del vqa_rad_train, slake_train, slake_val, vqa_rad_test, slake_test
gc.collect()

print(f"\nCOMBINÉ — Train: {len(combined_train)} | Val: {len(combined_val)} | Test: {len(combined_test)}")
closed_in_val = sum(1 for s in combined_val if s['answer_type'] == 'CLOSED')
print(f"Closed dans val : {closed_in_val} ({100*closed_in_val/len(combined_val):.1f}%)")
```

## 6. 📖 Step 3: Vocabulary Construction (Closed & Open)


```python
# =============================================================================
# ÉTAPE 3 : Vocabulaires
# =============================================================================
closed_counter = Counter(s['answer'] for s in combined_train if s['answer_type'] == 'CLOSED')
open_counter   = Counter(s['answer'] for s in combined_train if s['answer_type'] == 'OPEN')

TOP_CLOSED, TOP_OPEN = 500, 1000
top_closed = [ans for ans, _ in closed_counter.most_common(TOP_CLOSED)]
top_open   = [ans for ans, _ in open_counter.most_common(TOP_OPEN)]

closed2idx = {a: i for i, a in enumerate(top_closed)}
open2idx   = {a: i for i, a in enumerate(top_open)}
idx2closed = {i: a for a, i in closed2idx.items()}
idx2open   = {i: a for a, i in open2idx.items()}

NUM_CLOSED = len(closed2idx)
NUM_OPEN   = len(open2idx)
print(f"\nVocabulaires — Closed: {NUM_CLOSED} | Open: {NUM_OPEN}")
print(f"Top-5 Closed : {top_closed[:5]}")

def in_vocab(s):
    return s['answer'] in (closed2idx if s['answer_type'] == 'CLOSED' else open2idx)

combined_train = [s for s in combined_train if in_vocab(s)]
combined_val   = [s for s in combined_val   if in_vocab(s)]
combined_test  = [s for s in combined_test  if in_vocab(s)]
print(f"Après filtrage — Train: {len(combined_train)} | Val: {len(combined_val)} | Test: {len(combined_test)}")
```

## 7. 🖼️ Step 4: PyTorch Dataset, Data Augmentations & DataLoaders


```python
# =============================================================================
# ÉTAPE 4 : Dataset & Transforms
# =============================================================================
MEAN = [0.5, 0.5, 0.5]
STD  = [0.5, 0.5, 0.5]

train_transform = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.RandomCrop((224, 224)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=10),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=MEAN, std=STD),
])

val_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=MEAN, std=STD),
])

class MedVQADataset(Dataset):
    def __init__(self, data, transform, closed2idx, open2idx):
        self.data = data; self.transform = transform
        self.c2i = closed2idx; self.o2i = open2idx

    def __len__(self): return len(self.data)

    def __getitem__(self, idx):
        s = self.data[idx]
        image = Image.open(io.BytesIO(s['image_bytes'])).convert('RGB')
        image = self.transform(image)
        is_closed   = (s['answer_type'] == 'CLOSED')
        answer_type = 0 if is_closed else 1
        label = self.c2i[s['answer']] if is_closed else self.o2i[s['answer']]
        return image, s['question'], answer_type, label, s['answer']
```

## 8. 🔤 Step 5: BiomedBERT Tokenizer


```python
# =============================================================================
# ÉTAPE 5 : Tokenizer
# =============================================================================
BIOMEDBERT = "microsoft/BiomedNLP-BiomedBERT-base-uncased-abstract"
tokenizer  = AutoTokenizer.from_pretrained(BIOMEDBERT)

def collate_fn(batch):
    images, questions, ans_types, labels, answers = zip(*batch)
    images = torch.stack(images)
    tokens = tokenizer(list(questions), padding=True, truncation=True,
                       max_length=64, return_tensors='pt')
    return (images, tokens['input_ids'], tokens['attention_mask'],
            torch.tensor(ans_types, dtype=torch.long),
            torch.tensor(labels,    dtype=torch.long),
            list(answers))

BATCH_SIZE = 16
train_dataset = MedVQADataset(combined_train, train_transform, closed2idx, open2idx)
val_dataset   = MedVQADataset(combined_val,   val_transform,   closed2idx, open2idx)
test_dataset  = MedVQADataset(combined_test,  val_transform,   closed2idx, open2idx)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True,
                          collate_fn=collate_fn, num_workers=2, pin_memory=PIN_MEM)
val_loader   = DataLoader(val_dataset,   batch_size=BATCH_SIZE, shuffle=False,
                          collate_fn=collate_fn, num_workers=2, pin_memory=PIN_MEM)
test_loader  = DataLoader(test_dataset,  batch_size=BATCH_SIZE, shuffle=False,
                          collate_fn=collate_fn, num_workers=2, pin_memory=PIN_MEM)

print(f"\nLoaders — Train: {len(train_loader)} | Val: {len(val_loader)} | Test: {len(test_loader)}")
```

## 9. 🔍 Step 6: Visual Explainability (Grad-CAM for ViT)


```python
# =============================================================================
# ÉTAPE 6 : GRAD-CAM MAISON
# =============================================================================
class GradCAMViT:
    """Grad-CAM sur ViT — accroche sur visual_encoder.norm."""
    def __init__(self, model):
        self.model = model
        self.grads, self.acts = [], []
        self.hook_f = model.visual_encoder.norm.register_forward_hook(
            lambda m, i, o: self.acts.append(o))
        self.hook_b = model.visual_encoder.norm.register_full_backward_hook(
            lambda m, gi, go: self.grads.append(go[0]))

    def remove_hooks(self):
        self.hook_f.remove(); self.hook_b.remove()

    def generate(self, img_t, input_ids, attn_mask, answer_type=0):
        self.model.eval()
        self.grads, self.acts = [], []
        lc, lo = self.model(img_t, input_ids, attn_mask)
        score  = lc.max() if answer_type == 0 else lo.max()
        self.model.zero_grad()
        score.backward(retain_graph=True)
        weights = self.grads[0].mean(dim=1, keepdim=True)
        cam = F.relu((weights * self.acts[0]).sum(dim=-1)[:, 1:])
        B   = cam.size(0)
        cam = cam.view(B, 14, 14)
        cam = cam - cam.view(B, -1).min(dim=1)[0].view(B, 1, 1)
        cam = cam / (cam.view(B, -1).max(dim=1)[0].view(B, 1, 1) + 1e-8)
        return cam.detach()
```

## 10. 🧠 Step 7: MedVQA N6 Architecture


```python
# =============================================================================
# ÉTAPE 7 : MODÈLE N6
# Architecture : ViT-B/16 + BiomedBERT + concat + MLP
# (Cross-Attention et BAN retirés — trop de params pour 7k samples)
#
# Améliorations légères vs N5+ :
#   + Projection dimensionnelle avant concat (dim 768 → 512 chacun)
#     → réduit les params de fusion de 30% sans perte d'info
#   + GELU au lieu de ReLU (meilleur gradient sur données médicales)
#   + BatchNorm1d avant les têtes (stabilise l'entraînement)
# =============================================================================
class MedVQA_N6(nn.Module):
    def __init__(self, num_closed, num_open, dim=768, proj_dim=512):
        super().__init__()

        # Encodeurs (identiques N5+)
        self.visual_encoder = timm.create_model(
            'vit_base_patch16_224', pretrained=True, num_classes=0)
        self.text_encoder = AutoModel.from_pretrained(BIOMEDBERT)

        # Projection légère avant concat (768 → 512)
        # Réduit params fusion de 30% vs concat brut
        self.v_proj = nn.Sequential(nn.Linear(dim, proj_dim), nn.GELU(), nn.Dropout(0.1))
        self.q_proj = nn.Sequential(nn.Linear(dim, proj_dim), nn.GELU(), nn.Dropout(0.1))

        # Fusion : [v_proj ; q_proj] → MLP
        self.fusion = nn.Sequential(
            nn.Linear(proj_dim * 2, proj_dim),
            nn.GELU(),
            nn.Dropout(0.3),
            nn.Linear(proj_dim, proj_dim // 2),
            nn.GELU(),
            nn.Dropout(0.3),
        )

        # BatchNorm avant têtes (stabilise)
        self.bn = nn.BatchNorm1d(proj_dim // 2)

        # Têtes de réponse
        self.head_closed = nn.Linear(proj_dim // 2, num_closed)
        self.head_open   = nn.Sequential(
            nn.Linear(proj_dim // 2, proj_dim // 4),
            nn.GELU(),
            nn.Dropout(0.3),
            nn.Linear(proj_dim // 4, num_open),
        )

    def forward(self, img, input_ids, attn_mask):
        # Encodage visuel : CLS token moyen
        v = self.visual_encoder.forward_features(img)
        v = v.mean(dim=1)                                    # (B, 768)

        # Encodage textuel : token [CLS]
        q = self.text_encoder(
            input_ids=input_ids, attention_mask=attn_mask
        ).last_hidden_state[:, 0, :]                         # (B, 768)

        # Projection + concat + fusion
        fused = self.fusion(torch.cat([self.v_proj(v), self.q_proj(q)], dim=-1))
        fused = self.bn(fused)

        return self.head_closed(fused), self.head_open(fused)

# =============================================================================
# Initialisation
# =============================================================================
gc.collect()
if device.type == 'cuda': torch.cuda.empty_cache()

model = MedVQA_N6(num_closed=NUM_CLOSED, num_open=NUM_OPEN).to(device)
n_params = sum(p.numel() for p in model.parameters())
print(f"\nParams totaux N6 : {n_params:,}")
print(f"(N5+ était : ~196,949,993 — N6 : similaire, fusion plus légère)")
if device.type == 'cuda':
    used  = torch.cuda.memory_allocated() / 1e9
    total = torch.cuda.get_device_properties(0).total_memory / 1e9
    print(f"VRAM après chargement modèle : {used:.1f} / {total:.1f} GB")
```

## 11. 🚀 Step 8: Two-Phase Training (Freeze Encoders -> Fine-Tuning)


```python
# =============================================================================
# ÉTAPE 8 : Entraînement — 2 phases (même stratégie que N5+)
# =============================================================================
criterion = nn.CrossEntropyLoss(label_smoothing=0.1)

def make_optimizer(phase):
    """Reset complet à chaque phase (plus stable que add_param_group)."""
    head_params = [
        {'params': model.v_proj.parameters(),      'lr': 1e-3},
        {'params': model.q_proj.parameters(),      'lr': 1e-3},
        {'params': model.fusion.parameters(),      'lr': 1e-3},
        {'params': model.bn.parameters(),          'lr': 1e-3},
        {'params': model.head_closed.parameters(), 'lr': 1e-3},
        {'params': model.head_open.parameters(),   'lr': 1e-3},
    ]
    if phase == 1:
        return torch.optim.AdamW(head_params, weight_decay=1e-2)
    else:
        return torch.optim.AdamW(head_params + [
            {'params': model.visual_encoder.parameters(), 'lr': 1e-5},
            {'params': model.text_encoder.parameters(),   'lr': 1e-5},
        ], weight_decay=1e-2)

# Phase 1 : encodeurs gelés
for p in model.visual_encoder.parameters(): p.requires_grad = False
for p in model.text_encoder.parameters():   p.requires_grad = False

phase     = 1
optimizer = make_optimizer(phase)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode='max', factor=0.5, patience=3)
scaler    = GradScaler('cuda') if USE_AMP else None

best_val       = 0.0
patience_limit = 10
no_improve     = 0
EPOCHS         = 40

print("\n=== Phase 1 : Encodeurs GELÉS (epochs 1-5) ===")

for epoch in range(EPOCHS):

    # ── Phase 2 : dégel complet epoch 5 ─────────────────────────────────────
    if epoch == 5 and phase == 1:
        phase = 2
        print("\n>>> Phase 2 : Dégel complet (LR encodeurs 1e-5) <<<")
        for p in model.visual_encoder.parameters(): p.requires_grad = True
        for p in model.text_encoder.parameters():   p.requires_grad = True
        optimizer = make_optimizer(phase)
        scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
            optimizer, mode='max', factor=0.5, patience=3)
        gc.collect()
        if device.type == 'cuda': torch.cuda.empty_cache()

    # ── TRAIN ────────────────────────────────────────────────────────────────
    model.train()
    total_loss = 0
    correct_c, total_c = 0, 0
    correct_o, total_o = 0, 0

    for imgs, input_ids, attn_mask, ans_types, labels, _ in train_loader:
        imgs      = imgs.to(device,      non_blocking=PIN_MEM)
        input_ids = input_ids.to(device, non_blocking=PIN_MEM)
        attn_mask = attn_mask.to(device, non_blocking=PIN_MEM)
        ans_types = ans_types.to(device, non_blocking=PIN_MEM)
        labels    = labels.to(device,    non_blocking=PIN_MEM)

        optimizer.zero_grad()
        mask_c = (ans_types == 0)
        mask_o = (ans_types == 1)

        if USE_AMP:
            with autocast('cuda'):
                logits_c, logits_o = model(imgs, input_ids, attn_mask)
                loss = torch.tensor(0.0, device=device)
                if mask_c.any(): loss = loss + criterion(logits_c[mask_c], labels[mask_c])
                if mask_o.any(): loss = loss + criterion(logits_o[mask_o], labels[mask_o])
            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            scaler.step(optimizer); scaler.update()
        else:
            logits_c, logits_o = model(imgs, input_ids, attn_mask)
            loss = torch.tensor(0.0, device=device)
            if mask_c.any(): loss = loss + criterion(logits_c[mask_c], labels[mask_c])
            if mask_o.any(): loss = loss + criterion(logits_o[mask_o], labels[mask_o])
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()

        total_loss += loss.item()
        with torch.no_grad():
            if mask_c.any():
                correct_c += (logits_c[mask_c].argmax(1) == labels[mask_c]).sum().item()
                total_c   += mask_c.sum().item()
            if mask_o.any():
                correct_o += (logits_o[mask_o].argmax(1) == labels[mask_o]).sum().item()
                total_o   += mask_o.sum().item()

    # ── VALIDATION ───────────────────────────────────────────────────────────
    model.eval()
    val_c, tot_c = 0, 0
    val_o, tot_o = 0, 0

    with torch.no_grad():
        for imgs, input_ids, attn_mask, ans_types, labels, _ in val_loader:
            imgs      = imgs.to(device,      non_blocking=PIN_MEM)
            input_ids = input_ids.to(device, non_blocking=PIN_MEM)
            attn_mask = attn_mask.to(device, non_blocking=PIN_MEM)
            ans_types = ans_types.to(device, non_blocking=PIN_MEM)
            labels    = labels.to(device,    non_blocking=PIN_MEM)
            logits_c, logits_o = model(imgs, input_ids, attn_mask)
            mask_c = (ans_types == 0)
            mask_o = (ans_types == 1)
            if mask_c.any():
                val_c += (logits_c[mask_c].argmax(1) == labels[mask_c]).sum().item()
                tot_c += mask_c.sum().item()
            if mask_o.any():
                val_o += (logits_o[mask_o].argmax(1) == labels[mask_o]).sum().item()
                tot_o += mask_o.sum().item()

    # ── MÉTRIQUES ────────────────────────────────────────────────────────────
    train_acc_c = 100 * correct_c / total_c if total_c > 0 else 0
    train_acc_o = 100 * correct_o / total_o if total_o > 0 else 0
    v_acc_c     = 100 * val_c / tot_c       if tot_c   > 0 else 0
    v_acc_o     = 100 * val_o / tot_o       if tot_o   > 0 else 0
    v_comb = (v_acc_c + v_acc_o) / 2.0 if (v_acc_c > 0 and v_acc_o > 0) else max(v_acc_c, v_acc_o)

    scheduler.step(v_comb)

    if v_comb > best_val:
        best_val = v_comb
        torch.save({
            'epoch': epoch,
            'model_state_dict': model.state_dict(),
            'val_acc_closed': v_acc_c,
            'val_acc_open':   v_acc_o,
            'val_combined':   v_comb,
        }, 'best_n6.pth')
        saved = " ✓ SAVED"; no_improve = 0
    else:
        no_improve += 1
        saved = f" ({no_improve}/{patience_limit})"

    vram = f" | VRAM {torch.cuda.memory_allocated()/1e9:.1f}GB" if device.type == 'cuda' else ""
    print(f"Ep {epoch+1:02d} [Ph{phase}] | Loss {total_loss/len(train_loader):.4f} | "
          f"Train C:{train_acc_c:.1f}% O:{train_acc_o:.1f}% | "
          f"Val C:{v_acc_c:.1f}% O:{v_acc_o:.1f}% Comb:{v_comb:.1f}%"
          f"{saved}{vram}")

    if no_improve >= patience_limit:
        print(f"\nEarly stopping à l'epoch {epoch+1}")
        break

print(f"\nBest Val Combined: {best_val:.1f}%")
```

## 12. 📊 Step 9: Comprehensive Test Evaluation (Accuracy, BLEU, ROUGE)


```python
# =============================================================================
# ÉTAPE 9 : TEST + MÉTRIQUES COMPLÈTES
# =============================================================================
gc.collect()
if device.type == 'cuda': torch.cuda.empty_cache()

ck = torch.load('best_n6.pth', map_location=device)
model.load_state_dict(ck['model_state_dict'])
model.eval()

all_preds_c, all_labels_c = [], []
all_preds_o, all_labels_o = [], []
all_pred_str_o, all_ref_str_o = [], []

with torch.no_grad():
    for imgs, input_ids, attn_mask, ans_types, labels, answers in test_loader:
        imgs      = imgs.to(device, non_blocking=PIN_MEM)
        input_ids = input_ids.to(device, non_blocking=PIN_MEM)
        attn_mask = attn_mask.to(device, non_blocking=PIN_MEM)
        logits_c, logits_o = model(imgs, input_ids, attn_mask)
        mask_c = (ans_types == 0)
        mask_o = (ans_types == 1)
        if mask_c.any():
            preds = logits_c[mask_c].argmax(1).cpu().tolist()
            all_preds_c.extend(preds)
            all_labels_c.extend(labels[mask_c].tolist())
        if mask_o.any():
            preds = logits_o[mask_o].argmax(1).cpu().tolist()
            all_preds_o.extend(preds)
            all_labels_o.extend(labels[mask_o].tolist())
            for p, ref in zip(preds, [answers[i] for i, m in enumerate(mask_o) if m]):
                all_pred_str_o.append(idx2open.get(p, '<unk>'))
                all_ref_str_o.append(ref)

closed_acc = np.mean(np.array(all_preds_c) == np.array(all_labels_c)) * 100 if all_preds_c else 0
open_acc   = np.mean(np.array(all_preds_o) == np.array(all_labels_o)) * 100 if all_preds_o else 0
f1_c = f1_score(all_labels_c, all_preds_c, average='macro', zero_division=0) * 100 if all_labels_c else 0
f1_o = f1_score(all_labels_o, all_preds_o, average='macro', zero_division=0) * 100 if all_labels_o else 0
bleu1 = 0.0
if all_pred_str_o:
    bleu_metric = BLEU(max_ngram_order=1)
    bleu1 = bleu_metric.corpus_score(all_pred_str_o, [[r] for r in all_ref_str_o]).score
vqa_score = (closed_acc * len(all_preds_c) + open_acc * len(all_preds_o)) / \
            max(len(all_preds_c) + len(all_preds_o), 1)

print("\n" + "=" * 65)
print("  TEST RESULTS — N6 Final (ViT + BiomedBERT + proj + concat)")
print("=" * 65)
print(f"  Closed Accuracy  : {closed_acc:.1f}%   (N5+: ~65% | N3: 57.35%)")
print(f"  Open   Accuracy  : {open_acc:.1f}%   (N5+: ~77% | N3: 39.76%)")
print(f"  F1-macro Closed  : {f1_c:.1f}%")
print(f"  F1-macro Open    : {f1_o:.1f}%")
print(f"  BLEU-1 (open)    : {bleu1:.1f}")
print(f"  VQA Score        : {vqa_score:.1f}%")
print("=" * 65)

if closed_acc > 70 and open_acc > 77:
    print("  ✅ Gain confirmé vs N5+ — architecture N6 validée")
elif closed_acc > 64:
    print("  ✅ Résultats stables — amélioration marginale vs N5+")
else:
    print("  ⚠️  Résultats similaires N5+ — val stratifiée seule a aidé")
```

```python
# =============================================================================
# ÉTAPE 11 : GRAD-CAM DÉMO
# =============================================================================
print("\n=== Grad-CAM Démo (5 exemples) ===")
try:
    import matplotlib
    matplotlib.use('Agg')
    import matplotlib.pyplot as plt

    gcam = GradCAMViT(model)
    fig, axes = plt.subplots(2, 5, figsize=(20, 8))
    shown = 0

    for i in range(len(test_dataset)):
        if shown >= 5: break
        img_t, question, ans_type, label, answer_str = test_dataset[i]
        img_b = img_t.unsqueeze(0).to(device)
        tok   = tokenizer([question], padding=True, truncation=True,
                          max_length=64, return_tensors='pt')
        iids  = tok['input_ids'].to(device)
        amask = tok['attention_mask'].to(device)

        with torch.no_grad():
            lc, lo = model(img_b, iids, amask)
        pred     = lc.argmax(1).item() if ans_type == 0 else lo.argmax(1).item()
        pred_str = idx2closed.get(pred, '?') if ans_type == 0 else idx2open.get(pred, '?')

        cam    = gcam.generate(img_b, iids, amask, ans_type)
        cam_np = cam[0].cpu().numpy()
        cam_up = np.array(
            Image.fromarray((cam_np * 255).astype(np.uint8)).resize((224, 224), Image.BILINEAR)
        ) / 255.0

        img_np = np.clip(img_t.permute(1, 2, 0).cpu().numpy() * np.array(STD) + np.array(MEAN), 0, 1)

        axes[0, shown].imshow(img_np)
        axes[0, shown].set_title(f"Q: {question[:35]}\nTrue: {answer_str}", fontsize=7)
        axes[0, shown].axis('off')
        axes[1, shown].imshow(img_np)
        axes[1, shown].imshow(cam_up, cmap='jet', alpha=0.5)
        axes[1, shown].set_title(f"Grad-CAM | Pred: {pred_str}", fontsize=7)
        axes[1, shown].axis('off')
        shown += 1

    gcam.remove_hooks()
    plt.suptitle("N6 Final — Grad-CAM (ViT + BiomedBERT)", fontsize=13, fontweight='bold')
    plt.tight_layout()
    plt.savefig('gradcam_n6.png', dpi=150, bbox_inches='tight')
    print("Grad-CAM sauvegardé : gradcam_n6.png")
    from IPython.display import display, Image as IPImage
    display(IPImage('gradcam_n6.png'))

except Exception as e:
    import traceback
    print(f"Grad-CAM erreur : {e}"); traceback.print_exc()
```
